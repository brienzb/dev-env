## MySQL Query

```python
import os
import pymysql
import pandas as pd

mysql_conn = pymysql.connect(
    host=LOCAL_HOST,
    user=os.getenv("MYSQL_USER"),
    password=os.getenv("MYSQL_PASSWORD"),
    database=os.getenv("MYSQL_DATABASE"),
    port=3306,
    charset="utf8mb4",
    cursorclass=pymysql.cursors.DictCursor,
)
```

```python
cursor = mysql_conn.cursor()

query = """
select *
from test_table
"""

cursor.execute(query)
result = cursor.fetchall()

df = pd.DataFrame(result)
df
```

```python
-- SQL Example
show tables;

create table test_table (
    no int(11) not null auto_increment,
    name varchar(20) not null,
    email varchar(100),
    gender char(1),
    age int(3),
    constraint test_table_pk primary key (no)
);

describe test_table;

insert into test_table value (null, 'user1', 'user1@example.com', 'M', 26);
insert into test_table value (null, 'user2', 'user2@example.com', 'M', 35);
insert into test_table value (null, 'user3', 'user3@example.com', 'F', 17);

select *
from test_table;
```

## MongoDB Query

```python
import os
import pymongo
import pandas as pd

mongo_client = pymongo.MongoClient(f"mongodb://{os.getenv('MONGO_INITDB_ROOT_USERNAME')}:{os.getenv('MONGO_INITDB_ROOT_PASSWORD')}@{LOCAL_HOST}:27017/")
db = mongo_client[os.getenv("MONGO_INITDB_DATABASE")]
collection = db["test_table"]
```

```python
data = collection.find({})

df = pd.DataFrame(list(data))
df
```

## Redis Query

```python
import json
import redis
import pandas as pd

redis_conn = redis.Redis(
    host=LOCAL_HOST,
    port=6379,
    decode_responses=True,
)
```

```python
keys = redis_conn.keys("*")
data = redis_conn.mget(keys)

data = [json.loads(item) for item in data]

df = pd.DataFrame(data)
df
```
